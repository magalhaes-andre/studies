```python
import pandas as pd

measurements = pd.read_csv("resources/b2w.csv")

measurements.head()
```

```python
clean_measurements = measurements.drop(axis=1, labels=["original_index", "review_text_processed", "review_text_tokenized", "kfold_polarity", "kfold_rating", "rating"], )

clean_measurements.head()
```

```python
clean_measurements.dropna(inplace=True, axis=0)
clean_measurements.head()
```

```python
clean_measurements.polarity.value_counts()
```

```python
from sklearn.model_selection import train_test_split

train, test, train_class, test_class = train_test_split(
    measurements.review_text, 
    measurements.polarity,
    stratify=measurements.polarity,
    random_state=71)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(train, train_class)
accuracy = lr.score(test, test_class)
accuracy
```
